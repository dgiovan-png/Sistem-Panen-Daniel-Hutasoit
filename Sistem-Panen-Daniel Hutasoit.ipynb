```python
# Sel 6: Perbarui file sistem_panen.py dengan fungsi diskon
%%writefile sistem_panen.py
def hitung_total_panen(berat_kg):
    harga_per_kg = 15000
    return berat_kg * harga_per_kg

def hitung_diskon(total_harga):
    if total_harga >= 1000000:
        return total_harga * 0.10  # Diskon 10% jika >= 1 Juta
    return 0

print("=== SISTEM PENCATATAN HASIL PANEN ===")
nama = input("Nama Petani: ")
berat = float(input("Masukkan Berat Panen (kg): "))

total = hitung_total_panen(berat)
diskon = hitung_diskon(total)
total_akhir = total - diskon

print(f"Total Pendapatan Sebelum Diskon: Rp {total:,.0f}")
print(f"Diskon: Rp {diskon:,.0f}")
print(f"Total Pendapatan Akhir {nama}: Rp {total_akhir:,.0f}")
```
