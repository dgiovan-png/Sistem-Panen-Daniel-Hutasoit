# Penyelesaian Merge Conflict

Versi akhir disamakan dengan coding Daniel pada main branch.
Harga panen yang digunakan adalah Rp18.000/kg.

```python
def hitung_total_panen(berat_kg):
    # Resolusi conflict: samakan dengan main branch Daniel
    harga_per_kg = 18000
    return berat_kg * harga_per_kg

# Contoh hasil setelah conflict diselesaikan
berat = 10
total = hitung_total_panen(berat)
print(f'{berat} kg x Rp{18000:,.0f} = Rp{total:,.0f}')
```
