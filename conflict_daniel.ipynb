# Resolusi Konflik Harga Panen

Harga panen disetujui sesuai main branch Daniel, yaitu Rp18.000/kg.
Versi ini sudah disamakan agar tidak terjadi conflict saat merge.

```python
def hitung_total_panen(berat_kg):
    # Resolusi konflik: harga final = Rp18.000/kg
    harga_per_kg = 18000
    return berat_kg * harga_per_kg

berat = 10
total = hitung_total_panen(berat)
print(f'{berat} kg x Rp{18000:,.0f} = Rp{total:,.0f}')
```
