# Penyelesaian Merge Conflict

Daniel menetapkan harga Rp18.000/kg, sedangkan Fazel menetapkan Rp20.000/kg.
Resolusi yang dipilih untuk branch Fazel adalah Rp20.000/kg.

```python
def hitung_total_panen(berat_kg):
    # Resolusi conflict: gunakan harga Fazel Rp20.000/kg
    harga_per_kg = 20000
    return berat_kg * harga_per_kg

# Contoh hasil setelah conflict diselesaikan
berat = 10
total = hitung_total_panen(berat)
print(f'{berat} kg x Rp{20000:,.0f} = Rp{total:,.0f}')
```
